## song part indicator removal

```python
import re
def remove_feature(text):
    return re.sub(r'\[.*?\]', '', text).strip()

df['lyrics'] = df['lyrics'].apply(remove_feature)

for text in df['lyrics'][3374193:]:
    print(text)
```

## feature cleaned

```python
df['features'][6]
```

```python
import re
def feature_cleaned(text):
    cleaned_text = re.sub(r'[{}\'"]', '', text)  
    cleaned_text = re.sub(r',', ' ', cleaned_text) 
    return cleaned_text.strip()

df['features'] = df['features'].apply(feature_cleaned)

for text in df['features'][3374193:]:
    print(text)
```

```python
df['song_document'] = df['title'].fillna("") + " " + df['lyrics_cleaned'].fillna("") + " " + df['feature_cleaned'].fillna("")
```
